# Notebook: Import CSV Data into Delta Lake Tables

## Step 1: Import the data from the GitHub Repository

````python
import os, requests

# Download demo data files into lakehouse if not exist
remote_url = "https://raw.githubusercontent.com/hgrandy94/health_insurance_claims/refs/heads/main/data"

file_list = [
    "claim_investigations",
    "claims",
    "policyholders"
] 

file_path = "/lakehouse/default/Files"

if not os.path.exists("/lakehouse/default"):
    raise FileNotFoundError(
        "Default lakehouse not found, please add a lakehouse and restart the session."
    )
for fname in file_list:
    if not os.path.exists(f"{file_path}/{fname}.csv"):
        r = requests.get(f"{remote_url}/{fname}.csv", timeout=30)
        r.raise_for_status()
        with open(f"{file_path}/{fname}.csv", "wb") as f:
            f.write(r.content)
print("Downloaded demo data files into lakehouse.")
````

## Step 2: Convert CSV files into Managed Delta-Parquet Tables

````python
# Read raw data from the Files section of the lakehouse and convert it into delta tables. These tables will then appear in the Tables section of the lakehouse

for fname in file_list:
    try:
        spark.read.option("header", True).option("inferSchema", True).csv(f"Files/{fname}.csv").write.mode("overwrite").format("delta").saveAsTable(fname)
    except Exception as e:
        print(f"Error --> {e}")
````
